```python
from transformers import BertConfig, BertModel
#直接饮用会胡说八道
config = BertConfig()
model = BertModel(config)

# Model is randomly initialized!

from transformers import BertTokenizer, BertModel

# 1. 指定一个预训练好的模型名称
# "bert-base-uncased" 是一个常用的、预训练好的 BERT 模型
model_name = "bert-base-uncased"

# 2. 从预训练模型加载分词器（Tokenizer）
# Tokenizer 负责将你的文本转换成模型能理解的数字ID
tokenizer = BertTokenizer.from_pretrained(model_name)

# 3. 从预训练模型加载模型（Model）
# 这会自动下载并加载预训练好的权重，而不是随机初始化
model = BertModel.from_pretrained(model_name)

# 4. 准备你的输入文本
text = "The quick brown fox jumps over the lazy dog."

# 5. 使用 tokenizer 处理输入
# `return_tensors="pt"` 会返回 PyTorch 张量
inputs = tokenizer(text, return_tensors="pt")

# 6. 将处理好的输入传递给模型，获取输出
# 模型会返回一个包含隐藏状态（hidden states）等信息的字典
outputs = model(**inputs)

# 7. 获取隐藏状态
# last_hidden_state 是模型输出的最后一个隐藏层
last_hidden_states = outputs.last_hidden_state

print("Input shape:", inputs['input_ids'].shape)
print("Output shape:", last_hidden_states.shape)
print("Model output (truncated):", last_hidden_states)
```

## step2 模型可理解的内容

要将文本转换为模型可以理解的数字（即词汇表索引），你需要使用分词器（Tokenizer）。分词器是 Transformer 模型不可或缺的一部分，它负责将原始文本字符串处理成一系列数字，这些数字代表着模型词汇表中的特定词语或子词。

```python
sequences = ["Hello!", "Cool.", "Nice!"]
```

```python

# 加载预训练好的分词器
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
```

```python
tokenized_sequences = tokenizer(sequences)

print(tokenized_sequences)
```

### step2.2 基于子词（subword）的 tokenization

AutoModel 和 AutoTokenizer：是抽象的工厂类，能根据模型名称自动帮你加载任何类型的模型和分词器，极大地提高了代码的通用性和便捷性。你可以用同样的代码来加载不同的模型，而不需要更改类名。
